This line plot below shows a positive relationship between tournament wins and average regular-season win percentage. Teams with more tournament wins generally had higher regular-season win percentages. This suggests that regular-season performance may be useful in predicting how many tournament games a team wins.

```r
tidydata |>
  group_by(tourney_wins) |>
  summarise(mean_pct = mean(reg_wins_pct, na.rm = TRUE)) |>
  ggplot(aes(x = tourney_wins, y = mean_pct)) +
  geom_line(color = "steelblue") +
  geom_point(color = "steelblue", size = 2) +
  labs(
    title = "Average Regular Season Performance by Tournament Wins",
    x = "Tournament Wins",
    y = "Average Regular Season Win Percentage"
  )
```

The boxplot below compares the number of tournament wins between teams that played their first tournament game at home and those that did not. Teams playing at home had a higher median of 2 tournament wins compared to 0 for teams playing away. This suggests that playing the first game at home may be useful in predicting tournament success.

```r
ggplot(tidydata, aes(x = first_game_at_home,
                     y = tourney_wins,
                     fill = first_game_at_home)) +
  geom_boxplot() +
  labs(
    title = "Tournament Wins by First Game Location",
    x = "First Game at Home (N = No, Y = Yes)",
    y = "Tournament Wins"
  )
```
